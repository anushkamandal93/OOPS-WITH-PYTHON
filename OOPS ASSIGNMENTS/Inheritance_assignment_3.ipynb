# Q1. Electronics and Laptop Class

## Problem Statement
Create a parent class `Electronics` and child class `Laptop`.

Store:
- brand_name
- processor

Display all details using method.

## Sample Input
```python
Brand: Dell
Processor: i7
```

## Sample Output
```python
Brand: Dell
Processor: i7
```


````python
class Electronics:

  def __init__(self, brand_name):
    self.brand_name = brand_name


class Laptop(Electronics):

  def __init__(self, brand_name, processor):
    super().__init__(brand_name)
    self.processor = processor

  def display(self):
    print(f"Brand : {self.brand_name} Processor : {self.processor}")


# Example usage matching output:
laptop1 = Laptop("Dell", "i7")
laptop1.display()
````

```python
Country: India
```


# Q5. Father and Mother Multiple Inheritance

## Problem Statement
Create:
- Father class
- Mother class
- Son class

Store:
- father_profession
- mother_profession
- son_name

Display all details.

## Sample Input
```python
Father Profession: Engineer
Mother Profession: Teacher
Son Name: Rahul
```

## Sample Output
```python
Father Profession: Engineer
Mother Profession: Teacher
Son Name: Rahul
```


````python
class Father:

  def __init__(self, father_profession):
    self.father_profession = father_profession


class Mother:

  def __init__(self, mother_profession):
    self.mother_profession = mother_profession


class Son(Father, Mother):

  def __init__(self, father_profession, mother_profession, son_name):
    Father.__init__(self, father_profession)
    Mother.__init__(self, mother_profession)
    self.son_name = son_name

  def display(self):
    print(
        f"Father Profession : {self.father_profession} Mother Profession :"
        f" {self.mother_profession} Son Name : {self.son_name}"
    )


# Example usage matching output:
son1 = Son("Engineer", "Teacher", "Rahul")
son1.display()
````


# Q10. Teacher Hierarchical Inheritance

## Problem Statement
Create hierarchical inheritance using:
- Teacher
- MathsTeacher
- ScienceTeacher

Store:
- teacher_name
- subject

Display details using both child classes.

## Sample Input
```python
Teacher: Neha
Subject: Maths

Teacher: Rahul
Subject: Science
```

## Sample Output
```python
Teacher: Neha
Subject: Maths

Teacher: Rahul
Subject: Science
```


````python
class Teacher:

  def __init__(self, teacher_name, subject):
    self.teacher_name = teacher_name
    self.subject = subject


class MathsTeacher(Teacher):

  def display(self):
    print(f"Teacher : {self.teacher_name} Subject : {self.subject}")


class ScienceTeacher(Teacher):

  def display(self):
    print(f"Teacher : {self.teacher_name} Subject : {self.subject}")


# Example usage matching output:
math_t = MathsTeacher("Neha", "Maths")
sci_t = ScienceTeacher("Rahul", "Science")
math_t.display()
sci_t.display()
````
