# Q1. Create a Player Class

## Problem Statement
Create a class named `Player` using constructor.

Store:
- Player Name
- Jersey Number

Create a method to display player details.

Create two objects.

## Sample Output
```python
Player Name: Virat
Jersey Number: 18

Player Name: Rohit
Jersey Number: 45
```


````python
class Player:

  def __init__(self, name, jersey_number):
    self.name = name
    self.jersey_number = jersey_number

  def display_details(self):
    print(f"Player Name : {self.name}")
    print(f"Jersey Number : {self.jersey_number}")


# Example usage matching output:
player1 = Player("Virat", 18)
player1.display_details()

player2 = Player("Rohit", 45)
player2.display_details()
````

# Q3. Employee Bonus Calculator

## Problem Statement
Create a class named `Employee`.

Store:
- Employee Name
- Salary

Create method to calculate bonus:
- Bonus = 10% of salary

## Sample Output
```python
Employee Name: Aman
Salary: 50000
Bonus: 5000
```


````python
class Employee:

  def __init__(self, name, salary):
    self.name = name
    self.salary = salary

  def calculate_bonus(self):
    bonus = int(self.salary * 0.10)
    print(f"Employee Name : {self.name}")
    print(f"Salary : {self.salary}")
    print(f"Bonus : {bonus}")


# Example usage matching output:
emp1 = Employee("Aman", 50000)
emp1.calculate_bonus()
````

# Q4. Circle Class

## Problem Statement
Create a class named `Circle`.

Use constructor to take radius.

Create methods to calculate:
- Area
- Circumference

Use:
```python
3.14 * r * r
2 * 3.14 * r
```

## Sample Output
```python
Area: 78.5
Circumference: 31.400000000000002
```


````python
class Circle:

  def __init__(self, radius):
    self.radius = radius

  def calculate_area(self):
    return 3.14 * self.radius * self.radius

  def calculate_circumference(self):
    return 2 * 3.14 * self.radius


# Example usage matching output:
circle1 = Circle(5)
print(f"Area : {circle1.calculate_area()}")
print(f"Circumference : {circle1.calculate_circumference()}")
````

# Q7. Bank Account Class

## Problem Statement
Create a class named `BankAccount`.

Store:
- Account Holder Name
- Balance

Create method `deposit()` to add money into balance.

## Sample Output
```python
Name: Arjun
Balance After Deposit: 55000
```


````python
class BankAccount:

  def __init__(self, account_holder, balance):
    self.account_holder = account_holder
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount
    print(f"Name : {self.account_holder}")
    print(f"Balance After Deposit : {self.balance}")


# Example usage matching output:
account = BankAccount("Arjun", 50000)
account.deposit(5000)
````

# Q10. Food Item Class

## Problem Statement
Create a class named `FoodItem`.

Store:
- Food Name
- Price
- Quantity

Create method to calculate total bill.

Formula:
```python
Total = Price * Quantity
```

## Sample Output
```python
Food Name: Pizza
Total Bill: 800
```


````python
class FoodItem:

  def __init__(self, food_name, price, quantity):
    self.food_name = food_name
    self.price = price
    self.quantity = quantity

  def calculate_total_bill(self):
    total = int(self.price * self.quantity)
    print(f"Food Name : {self.food_name}")
    print(f"Total Bill : {total}")


# Example usage matching output:
food = FoodItem("Pizza", 400, 2)
food.calculate_total_bill()
````
