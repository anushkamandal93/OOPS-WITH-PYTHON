Multilevel Inheritance – Employee Hierarchy
  
    Problem Statement

    Create:

    Person
    Employee
    Manager

    Display all employee details.
    
     Output:
     Name : Amit
     Salary : 50000
     Department : IT

```python
class Person:

  def __init__(self, name):
    self.name = name


class Employee(Person):

  def __init__(self, name, salary):
    super().__init__(name)
    self.salary = salary


class Manager(Employee):

  def __init__(self, name, salary, department):
    super().__init__(name, salary)
    self.department = department

  def display(self):
    print(f"Name : {self.name}")
    print(f"Salary : {self.salary}")
    print(f"Department : {self.department}")


# Example usage matching output:
mgr = Manager("Amit", 50000, "IT")
mgr.display()
```

Using super() Function

    Problem Statement

    Create:

    Vehicle
    Car

    Use super() to access parent constructor.
    
    
    Output:
    Brand : Toyota
    Model : Fortuner

```python
class Vehicle:

  def __init__(self, brand):
    self.brand = brand


class Car(Vehicle):

  def __init__(self, brand, model):
    super().__init__(brand)
    self.model = model

  def display(self):
    print(f"Brand : {self.brand}")
    print(f"Model : {self.model}")


# Example usage matching output:
car = Car("Toyota", "Fortuner")
car.display()
```

Encapsulation – Bank Account

    Problem Statement

    Create a private balance variable and access it using methods.
    
    Output:
    Final Price = 45000.0

```python
class BankAccount:

  def __init__(self, balance):
    self.__balance = balance  # Private variable

  def set_balance(self, balance):
    self.__balance = balance

  def get_balance(self):
    return self.__balance


# Example usage matching output:
acc = BankAccount(45000.0)
print(f"Final Price = {acc.get_balance()}")
```

Hospital Management System
   
    Problem Statement

    Create:

    Person
    Doctor
    Patient

    Display all information.
    
    Output:
    Doctor : Sharma
    Specialization : Cardiologist

```python
class Person:

  def __init__(self, name):
    self.name = name


class Doctor(Person):

  def __init__(self, name, specialization):
    super().__init__(name)
    self.specialization = specialization

  def display(self):
    print(f"Doctor : {self.name}")
    print(f"Specialization : {self.specialization}")


class Patient(Person):

  def __init__(self, name, disease):
    super().__init__(name)
    self.disease = disease


# Example usage matching output:
doc = Doctor("Sharma", "Cardiologist")
doc.display()
```

Polymorphism – Shape Area

    Problem Statement

     Create:

     Rectangle
     Circle

     Both should have area() method.
     
     Output:
     Area = 50
     Area = 153.86

```python
class Rectangle:

  def __init__(self, length, width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width


class Circle:

  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return round(3.14 * self.radius * self.radius, 2)


# Example usage matching output:
rect = Rectangle(10, 5)
circle = Circle(7)
print(f"Area = {rect.area()}")
print(f"Area = {circle.area()}")
```

Online Shopping System
    
    Problem Statement

    Create:

    Product
    Electronics Product

  Apply discount calculation.
   
    Output: Final Price = 45000.0

```python
class Product:

  def __init__(self, price):
    self.price = price


class ElectronicsProduct(Product):

  def calculate_discounted_price(self):
    # Applying discount logic to match output
    discount = self.price * 0.10
    final_price = self.price - discount
    return final_price


# Example usage matching output:
item = ElectronicsProduct(50000.0)
print(f"Final Price = {item.calculate_discounted_price()}")
```
