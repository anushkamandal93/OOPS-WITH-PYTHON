# Q1. Animal Sound Program

## Problem Statement
Create an abstract class `Animal` with abstract method `sound()`.

Create child class `Dog` and print dog sound.

## Sample Input
```python
Dog
```

## Sample Output
```python
Dog barks
```


````python
from abc import ABC, abstractmethod


class Animal(ABC):

  @abstractmethod
  def sound(self):
    pass


class Dog(Animal):

  def sound(self):
    print("Dog barks")


# Example usage matching output:
d = Dog()
d.sound()
````


# Q3. Shape Area Program

## Problem Statement
Create abstract class `Shape` with abstract method `area()`.

Create child class `Square`.

Calculate square area.

## Formula
```python
side * side
```

## Sample Input
```python
Side: 5
```

## Sample Output
```python
Area: 25
```


````python
from abc import ABC, abstractmethod


class Shape(ABC):

  @abstractmethod
  def area(self):
    pass


class Square(Shape):

  def __init__(self, side):
    self.side = side

  def area(self):
    return self.side * self.side


# Example usage matching output:
sq = Square(5)
print(f"Area : {sq.area()}")
````


# Q10. Student Result Program

## Problem Statement
Create abstract class `Student` with abstract method `result()`.

Create child class `SchoolStudent`.

Print result status.

## Sample Input
```python
Marks: 80
```

## Sample Output
```python
Result: Pass
```


````python
from abc import ABC, abstractmethod


class Student(ABC):

  @abstractmethod
  def result(self, marks):
    pass


class SchoolStudent(Student):

  def result(self, marks):
    if marks >= 40:
      print("Result : Pass")
    else:
      print("Result : Fail")


# Example usage matching output:
student = SchoolStudent()
student.result(80)
````
