# Q1. Single Inheritance — Animal and Cat

## Problem Statement
Create a parent class `Animal` with:
- animal_name

Create child class `Cat` with:
- color

Create method to display all details.

## Sample Output
```python
Animal Name: Cat
Color: White
```


````python
class Animal:

  def __init__(self, animal_name):
    self.animal_name = animal_name


class Cat(Animal):

  def __init__(self, animal_name, color):
    super().__init__(animal_name)
    self.color = color

  def display(self):
    print(f"Animal Name : {self.animal_name} Color : {self.color}")


# Example usage matching sample output:
cat1 = Cat("Cat", "White")
cat1.display()
````


# Q2. Single Inheritance — Person and Employee

## Problem Statement
Create a parent class `Person` with:
- name
- age

Create child class `Employee` with:
- salary

Display all details using method.

## Sample Output
```python
Name: Rahul
Age: 28
Salary: 50000
```


````python
class Person:

  def __init__(self, name, age):
    self.name = name
    self.age = age


class Employee(Person):

  def __init__(self, name, age, salary):
    super().__init__(name, age)
    self.salary = salary

  def display(self):
    print(f"Name : {self.name} Age : {self.age} Salary : {self.salary}")


# Example usage matching sample output:
emp1 = Employee("Rahul", 28, 50000)
emp1.display()
````


# Q3. Multilevel Inheritance — Grandfather, Father, Son

## Problem Statement
Create:
- Grandfather class
- Father class inherited from Grandfather
- Son class inherited from Father

Store:
- grandfather_name
- father_name
- son_name

Display all details.

## Sample Output
```python
Grandfather: Mohan
Father: Rajesh
Son: Aman
```


````python
class Grandfather:

  def __init__(self, grandfather_name):
    self.grandfather_name = grandfather_name


class Father(Grandfather):

  def __init__(self, grandfather_name, father_name):
    super().__init__(grandfather_name)
    self.father_name = father_name


class Son(Father):

  def __init__(self, grandfather_name, father_name, son_name):
    super().__init__(grandfather_name, father_name)
    self.son_name = son_name

  def display(self):
    print(
        f"Grandfather : {self.grandfather_name} Father : {self.father_name}"
        f" Son : {self.son_name}"
    )


# Example usage matching sample output:
son1 = Son("Mohan", "Rajesh", "Aman")
son1.display()
````


# Q5. Multiple Inheritance — Father and Mother

## Problem Statement
Create:
- Father class
- Mother class

Create child class `Child` inherited from both classes.

Store:
- father_name
- mother_name
- child_name

Display all details.

## Sample Output
```python
Father: Raj
Mother: Sunita
Child: Ravi
```


````python
class Father:

  def __init__(self, father_name):
    self.father_name = father_name


class Mother:

  def __init__(self, mother_name):
    self.mother_name = mother_name


class Child(Father, Mother):

  def __init__(self, father_name, mother_name, child_name):
    Father.__init__(self, father_name)
    Mother.__init__(self, mother_name)
    self.child_name = child_name

  def display(self):
    print(
        f"Father : {self.father_name} Mother : {self.mother_name} Child :"
        f" {self.child_name}"
    )


# Example usage matching sample output:
child1 = Child("Raj", "Sunita", "Ravi")
child1.display()
````


# Q7. Hierarchical Inheritance — Vehicle, Car, Bike

## Problem Statement
Create parent class `Vehicle`.

Create two child classes:
- Car
- Bike

Store:
- company_name
- model_name

Display details using both child classes.

## Sample Output
```python
Car Company: BMW
Car Model: X5

Bike Company: Yamaha
Bike Model: R15
```


````python
class Vehicle:

  def __init__(self, company_name, model_name):
    self.company_name = company_name
    self.model_name = model_name


class Car(Vehicle):

  def display(self):
    print(f"Car Company : {self.company_name} Car Model : {self.model_name}")


class Bike(Vehicle):

  def display(self):
    print(f"Bike Company : {self.company_name} Bike Model : {self.model_name}")


# Example usage matching sample output:
car1 = Car("BMW", "X5")
bike1 = Bike("Yamaha", "R15")
car1.display()
bike1.display()
````


# Q9. Multilevel Inheritance — Device, Mobile, Smartphone

## Problem Statement
Create:
- Device class
- Mobile class inherited from Device
- Smartphone class inherited from Mobile

Store:
- device_type
- brand
- ram

Display all details.

## Sample Output
```python
Device: Electronics
Brand: Samsung
RAM: 8GB
```


````python
class Device:

  def __init__(self, device_type):
    self.device_type = device_type


class Mobile(Device):

  def __init__(self, device_type, brand):
    super().__init__(device_type)
    self.brand = brand


class Smartphone(Mobile):

  def __init__(self, device_type, brand, ram):
    super().__init__(device_type, brand)
    self.ram = ram

  def display(self):
    print(
        f"Device : {self.device_type} Brand : {self.brand} RAM : {self.ram}"
    )


# Example usage matching sample output:
phone = Smartphone("Electronics", "Samsung", "8 GB")
phone.display()
````


# Q10. Multiple Inheritance — Camera and MusicPlayer

## Problem Statement
Create:
- Camera class
- MusicPlayer class

Create child class `SmartPhone` inherited from both classes.

Store:
- camera_mp
- speaker_type
- mobile_name

Display all details.

## Sample Output
```python
Camera: 108MP
Speaker: Dolby
Mobile: OnePlus
```


````python
class Camera:

  def __init__(self, camera_mp):
    self.camera_mp = camera_mp


class MusicPlayer:

  def __init__(self, speaker_type):
    self.speaker_type = speaker_type


class SmartPhone(Camera, MusicPlayer):

  def __init__(self, camera_mp, speaker_type, mobile_name):
    Camera.__init__(self, camera_mp)
    MusicPlayer.__init__(self, speaker_type)
    self.mobile_name = mobile_name

  def display(self):
    print(
        f"Camera : {self.camera_mp} Speaker : {self.speaker_type} Mobile :"
        f" {self.mobile_name}"
    )


# Example usage matching sample output:
phone = SmartPhone("108 MP", "Dolby", "OnePlus")
phone.display()
````
