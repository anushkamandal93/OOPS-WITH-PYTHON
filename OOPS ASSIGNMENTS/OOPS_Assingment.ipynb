Q1. Student Information System
      
    Problem Statement

    Create a class named Student.

    Store:

      Student Name
      Roll Number
      Course

Create a method display_details() to display all student information.

       Output:
       Student Name : Rahul
       Roll Number  : 101
       Course       : Python

```python
class Student:

  def __init__(self, name, roll_number, course):
    self.name = name
    self.roll_number = roll_number
    self.course = course

  def display_details(self):
    print(f"Student Name : {self.name}")
    print(f"Roll Number  : {self.roll_number}")
    print(f"Course       : {self.course}")


# Example usage matching output:
student1 = Student("Rahul", 101, "Python")
student1.display_details()
```

Q3. Bank Account System

    Problem Statement

    Create a class named BankAccount.

    Store:

    Account Holder Name
    Balance

    Create methods:

     deposit()
     withdraw()
    check_balance()
    
    Output: Current Balance = 11000

```python
class BankAccount:

  def __init__(self, account_holder, balance=0):
    self.account_holder = account_holder
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount

  def withdraw(self, amount):
    if amount <= self.balance:
      self.balance -= amount
    else:
      print("Insufficient balance!")

  def check_balance(self):
    print(f"Current Balance = {self.balance}")


# Example usage matching output:
account = BankAccount("Rahul", 10000)
account.deposit(1000)
account.check_balance()
```

# ATM System

     Problem Statement

     Create a class named ATM.

     Store:

     Account Holder Name
     Balance

    Create methods:

    Deposit
    Withdraw
    Balance Check
    
     Output: Balance = 22000

```python
class ATM:

  def __init__(self, account_holder, balance=0):
    self.account_holder = account_holder
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount

  def withdraw(self, amount):
    if amount <= self.balance:
      self.balance -= amount
    else:
      print("Insufficient balance!")

  def balance_check(self):
    print(f"Balance = {self.balance}")


# Example usage matching output:
atm1 = ATM("Rahul", 20000)
atm1.deposit(2000)
atm1.balance_check()
```
